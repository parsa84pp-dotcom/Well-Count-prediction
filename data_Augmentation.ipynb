```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

np.random.seed(42)


INPUT_FILE = "data.csv"
OUTPUT_FILE = "aug_data.csv"
N_TARGET = 500      
K_NEIGHBORS = 5      



df = pd.read_csv(INPUT_FILE)
print("Data", df.shape)

feature_cols = df.columns.tolist()
X = df[feature_cols].values


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

nn = NearestNeighbors(n_neighbors=min(K_NEIGHBORS + 1, len(df)))
nn.fit(X_scaled)
_, indices = nn.kneighbors(X_scaled)


n_needed = N_TARGET - len(df)
synthetic_rows = []
for _ in range(n_needed):
    i = np.random.randint(0, len(df))
    neighbor_choices = indices[i][1:]          
    j = np.random.choice(neighbor_choices)
    lam = np.random.uniform(0, 1)
    new_row = X[i] + lam * (X[j] - X[i])       
    synthetic_rows.append(new_row)

df_synth = pd.DataFrame(synthetic_rows, columns=feature_cols)
df_aug = pd.concat([df, df_synth], ignore_index=True)


df_aug.to_csv(OUTPUT_FILE, index=False)
print(f"ُSaved : {OUTPUT_FILE}")
```
